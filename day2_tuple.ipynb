```python
import math

coordinates = (3, 4)
d0 = (0, 0)

def distance (x1, y1, x2, y2) :
    return math.sqrt(((x2 - x1)**(x2 - x1))+((y2 - y1)**(y2 - y1)))

resulte = distance(d0[0], d0[1], coordinates[0], coordinates[1])

print(resulte)
```
