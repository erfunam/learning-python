```python
#String interpolatin
esmam = "Seyed Erfan"
family = "Mohiyodin Jafari"
age = 20
print("Hello " + esmam + " " + family + " welcome")
```

```python
print("Hello %s %s welcome to %i" % (esmam, family, age))
```

```python
print("Hello {} {} you are {}".format(esmam, family, age))
```

```python
print("Hello {1} {0} you are {2}".format(esmam, family, age))
```

```python
print(f"Hello {esmam} {family} you are {age:1.1f}")
```
