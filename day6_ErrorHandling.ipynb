```python
def divis (a, b) :
    try :
        return a / b
    except ZeroDivisionError: 
        print (f"ZeroDivisionError")
    except Exception as e : 
            print (f"another erorr : {e}")

print (divis(3, [True,True]))
```

```python
def divis (a, b):
    try:
        r = a / b
    except Exception as ww:
        print(f"another erorr : {ww}")
    else:
        print(r)
    finally:
        return r

print (divis(5, 7))
```
