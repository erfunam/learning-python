```python
def is_even (n) :
    return (n % 2 == 0)

def any_even_in_list (numList) :
    for x in numList :
        if is_even(x) == True :
            return True
    return False

my_list = [1, 3, 6, 9, 7, 8]
print(any_even_in_list(my_list))
```

```python
def largest (num) :
    maxim = num[0]
    for lst_item in num :
        if maxim < lst_item :
            maxim = lst_item
    return maxim


my_list = [1, 3, 6, 9, 7, 8]
largest_number = largest(my_list)
print(largest_number)
```

```python
def is_even (n):
    return (n % 2 == 0)

def cList_odd (lst) :
    new_odd_list = []
    for items_list in lst :
        if not is_even(items_list):
            new_odd_list.append(items_list)
    return new_odd_list

my_list = [1, 3, 6, 9, 7, 8]
odd_res = cList_odd(my_list)
print(odd_res)
```

```python
def is_even (n):
    return (n % 2 == 0)

def cList_odd (lst) :
    new_odd_list = []
    counter = 0

    for items_list in lst :
        if not is_even(items_list):
            new_odd_list.append(items_list)
            counter += 1

    return counter, new_odd_list

my_list = [1, 3, 6, 9, 7, 8]
odds_count, odds_res = cList_odd(my_list)
print(odds_count, odds_res)
```
