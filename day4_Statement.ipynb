```python
numList = [1, 2, 12, 8, 7, 9, 6, 4]
lstZoj = []
lstFard = []
countZoj = 0
countFard = 0
for i in numList :
    if i % 2 == 0 :
        lstZoj.append(i)
        countZoj += 1
    else :
        lstFard.append(i)
        countFard += 1
    
    lstZoj.sort()
    lstFard.sort()

print(f"zoj ha : {lstZoj} tedad = {countZoj}")
print(f"fard ha : {lstFard} tedad = {countFard}")
```
